```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
df = pd.read_csv(r'F:\Pallavi\Learnings\Python\Modules\Deep_Learning\MLP\MLP_Regression\data\Exam_Score_Prediction.csv')
df
```

```python
df.drop(columns = ['student_id'], inplace = True)
x = df.drop(columns = ['exam_score'])
y = df.exam_score
encoded = pd.get_dummies(x, dtype = int)
encoded
```

```python
import pickle
feature_names = encoded.columns.tolist()
with open("feature_names.pkl", "wb") as f:
    pickle.dump(feature_names, f)
```

```python
x_train,x_test,y_train,y_test = train_test_split(encoded, y,test_size=0.2, random_state=10)
```

```python
model = Sequential()

model.add(Dense(units=8, activation='relu', input_dim=30))
model.add(Dense(units=4, activation='relu'))
model.add(Dense(units=1, activation='linear'))

model.compile(optimizer='adam', metrics=['r2_score'], loss='mean_squared_error')
history = model.fit(x_train, y_train, epochs=50, validation_split=0.2, batch_size=64,verbose=2)
```

```python
y_train_pred = model.predict(x_train)

mse = mean_squared_error(y_train, y_train_pred)
print("MSE : ",mse)
mae = mean_absolute_error(y_train, y_train_pred)
print("MAE : ", mae)
accuracy = r2_score(y_train, y_train_pred)
print("R2_Score", accuracy)
```

```python
y_test_pred = model.predict(x_test)

mse = mean_squared_error(y_test, y_test_pred)
print("MSE : ",mse)
mae = mean_absolute_error(y_test, y_test_pred)
print("MAE : ", mae)
accuracy = r2_score(y_test, y_test_pred)
print("R2_Score", accuracy)
```

```python
import keras_tuner as kt
```

```python
def build_Model(hp):
    model = Sequential()

    #Hiddlen layer 1
    unit_1 = hp.Int("unit_1", min_value= 8,max_value= 32, step =8)
    activ_fun1 = hp.Choice("activ_fun1", ["tanh","relu"])
    kernal_init1 = hp.Choice("kernal_init1",["he_normal","glorot_uniform"])
    model.add(Dense(units = unit_1, activation = activ_fun1, kernel_initializer=kernal_init1, input_dim=30))

    
     #Hiddlen layer 2
    unit_2 = hp.Int("unit_2", min_value= 4,max_value= 16, step =4)
    activ_fun2 = hp.Choice("activ_fun2", ["tanh","relu"])
    kernal_init2 = hp.Choice("kernal_init2",["he_normal","glorot_uniform"])
    model.add(Dense(units = unit_2, activation = activ_fun2, kernel_initializer=kernal_init2))

    #Output Layer
    model.add(Dense(units=1, activation="linear"))

    model.compile(optimizer="adam", metrics=["r2_score"],loss="mean_squared_error")

    return model

k_tuner = kt.RandomSearch(build_Model, max_trials= 4, objective="val_loss", project_name = "Proj1" )    
k_tuner.search(x_train,y_train, epochs=30, validation_data = (x_test,y_test), batch_size=256,verbose=2)

```

```python
import keras

def build_Model(hp):
    model = Sequential()

    #Hiddlen layer 1
    unit_1 = hp.Int("unit_1", min_value= 8,max_value= 32, step =8)
    activ_fun1 = hp.Choice("activ_fun1", ["tanh","relu"])
    kernal_init1 = hp.Choice("kernal_init1",["he_normal","glorot_uniform"])
    model.add(Dense(units = unit_1, activation = activ_fun1, kernel_initializer=kernal_init1, input_dim=30))

    
     #Hiddlen layer 2
    unit_2 = hp.Int("unit_2", min_value= 4,max_value= 16, step =4)
    activ_fun2 = hp.Choice("activ_fun2", ["tanh","relu"])
    kernal_init2 = hp.Choice("kernal_init2",["he_normal","glorot_uniform"])
    model.add(Dense(units = unit_2, activation = activ_fun2, kernel_initializer=kernal_init2))

    #Output Layer
    model.add(Dense(units=1, activation="linear"))

    #Optimizer
    opt_choice = hp.Choice("opt_choice",["adam","sgd","rmsprop"])
    learning_rate = hp.Float("learning_rate", min_value = 0.001, max_value=0.1, step = 0.005)

    if opt_choice=="adam":
        optimizer = keras.optimizers.Adam(learning_rate = learning_rate)
    elif opt_choice=="sgd":
        optimizer = keras.optimizers.SGD(learning_rate = learning_rate)
    elif opt_choice=="rmsprop":
        optimizer = keras.optimizers.RMSprop(learning_rate = learning_rate)

    model.compile(optimizer=optimizer, metrics=["r2_score"],loss="mean_squared_error")

    return model

k_tuner = kt.RandomSearch(build_Model, max_trials= 4, objective="val_loss", project_name = "Proj1" )    
k_tuner.search(x_train,y_train, epochs=30, validation_data = (x_test,y_test), batch_size=256,verbose=2)
```

```python
from tensorflow.keras.layers import BatchNormalization
```

```python
import keras

def build_Model(hp):
    model = Sequential()

    #Hiddlen layer 1
    unit_1 = hp.Int("unit_1", min_value= 8,max_value= 32, step =8)
    activ_fun1 = hp.Choice("activ_fun1", ["tanh","relu"])
    kernal_init1 = hp.Choice("kernal_init1",["he_normal","glorot_uniform"])
    model.add(Dense(units = unit_1, activation = activ_fun1, kernel_initializer=kernal_init1, input_dim=30))

    model.add(BatchNormalization(beta_initializer='zeros', gamma_initializer='ones'))

    
     #Hiddlen layer 2
    unit_2 = hp.Int("unit_2", min_value= 4,max_value= 16, step =4)
    activ_fun2 = hp.Choice("activ_fun2", ["tanh","relu"])
    kernal_init2 = hp.Choice("kernal_init2",["he_normal","glorot_uniform"])
    model.add(Dense(units = unit_2, activation = activ_fun2, kernel_initializer=kernal_init2))

    model.add(BatchNormalization(beta_initializer='zeros', gamma_initializer='ones'))

    #Output Layer
    model.add(Dense(units=1, activation="linear"))

    #Optimizer
    opt_choice = hp.Choice("opt_choice",["adam","sgd","rmsprop"])
    learning_rate = hp.Float("learning_rate", min_value = 0.001, max_value=0.1, step = 0.005)

    if opt_choice=="adam":
        optimizer = keras.optimizers.Adam(learning_rate = learning_rate)
    elif opt_choice=="sgd":
        optimizer = keras.optimizers.SGD(learning_rate = learning_rate)
    elif opt_choice=="rmsprop":
        optimizer = keras.optimizers.RMSprop(learning_rate = learning_rate)

    model.compile(optimizer=optimizer, metrics=["r2_score"],loss="mean_squared_error")

    return model

k_tuner = kt.RandomSearch(build_Model, max_trials= 4, objective="val_loss", project_name = "Proj1" )    
k_tuner.search(x_train,y_train, epochs=30, validation_data = (x_test,y_test), batch_size=256,verbose=2)
```
